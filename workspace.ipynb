# BookNooks dataset workspace

Run the cell below to load the newest collected CSV and display the first 50 book records as a dataframe. `books_50` keeps every field from the source file.

```python
from pathlib import Path
import pandas as pd
from IPython.display import display

# Works when launched from the project folder or from this notebook's folder.
project_dir = Path.cwd()
if not (project_dir / 'data' / 'processed').exists():
    project_dir = Path.cwd() / 'BookNooks'
    if not (project_dir / 'data' / 'processed').exists():
        project_dir = Path.cwd().parent

processed_dir = project_dir / 'data' / 'processed'
csv_files = list(processed_dir.glob('books_sample_*.csv'))
if not csv_files:
    raise FileNotFoundError(f'No collected books CSV found in {processed_dir}')

latest_csv = max(csv_files, key=lambda path: path.stat().st_mtime)
books_df = pd.read_csv(latest_csv)
books_50 = books_df.head(50).copy()

# Keep the full source descriptions in books_50, but show a readable
# excerpt in the displayed dataframe.
display_df = books_50.drop(
    columns=['rights_notes', 'description_word_count'], errors='ignore'
).copy()
def description_preview(text, max_words=40):
    words = str(text or '').split()
    preview = ' '.join(words[:max_words])
    return preview + ('…' if len(words) > max_words else '')

if 'description' in display_df.columns:
    display_df['description'] = display_df['description'].map(description_preview)

print(f'Source: {latest_csv.relative_to(project_dir)}')
print(f'Displaying {len(books_50)} of {len(books_df)} collected records')
display(display_df)
```
